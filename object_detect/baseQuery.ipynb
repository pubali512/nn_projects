```python
import sys
import subprocess
import importlib
import platform

def setup_detection_metric():
    # 1️⃣ Determine OS
    os_name = platform.system()
    
    # 2️⃣ Install correct pycocotools for OS
    try:
        import pycocotools
        print("✅ pycocotools already installed")
    except ModuleNotFoundError:
        pkg = "pycocotools-windows" if os_name == "Windows" else "pycocotools"
        print(f"Installing {pkg}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg])

    # 3️⃣ Install faster-coco-eval as fallback
    try:
        import faster_coco_eval
        print("✅ faster-coco-eval already installed")
    except ModuleNotFoundError:
        print("Installing faster-coco-eval...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "faster-coco-eval"])

    # 4️⃣ Monkey-patch TorchMetrics to see backends
    try:
        import pycocotools
        sys.modules["pycocotools"] = pycocotools
    except ModuleNotFoundError:
        pass
    try:
        import faster_coco_eval
        sys.modules["faster_coco_eval"] = faster_coco_eval
    except ModuleNotFoundError:
        pass

    # 5️⃣ Import MeanAveragePrecision
    try:
        from torchmetrics.detection.mean_ap import MeanAveragePrecision
        metric = MeanAveragePrecision()
        print("✅ TorchMetrics MeanAveragePrecision metric is ready to use!")
        return metric
    except ModuleNotFoundError as e:
        print("❌ Error importing TorchMetrics mAP metric:", e)
        return None

# Run setup
map_metric = setup_detection_metric()

















```
